```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
columns = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width',
    'species'
]

df = pd.read_csv(
    '/content/iris.data',
    header=None,
    names=columns
)

print("Dataset loaded successfully!")
print(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

df = df.dropna()

df = df.drop_duplicates()

print("\nDataset after cleaning:")
print(df.shape)

print("Data cleaning completed!")
```

```python
analysis = df.groupby("species").mean(numeric_only=True)

print("Average measurements by species:")
print(analysis)

print("\nFlower species count:")
print(df["species"].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="species")

plt.title("Distribution of Iris Flower Species")
plt.xlabel("Flower Species")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="sepal_width",
    hue="species",
    s=80
)

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

Findings and Observations

1. The Iris dataset contains three flower species: Setosa, Versicolor and Virginica.
2. Each species contains 50 flower records.
3. The dataset contains four numerical features: sepal length, sepal width, petal length and petal width.
4. The scatter plot shows differences in flower measurements between species.
5. Data visualization helps identify patterns in the dataset.
